### 02.3.2. Confirmando os insights encontrados a partir do relatório 

```python
columns = ['experiences_offered']

for i in columns:

    df_agg = fn_stats_pandas.groupby_count(df, i)

    print(f'- Agrupamento da coluna: {i}')

    display(HTML(df_agg.to_html(index = False)))

    fn_charts_pandas.barplot_donutplot(df_agg, i, 'Count')

    plt.savefig(f'../images/outputs/charts/pandas/nb02_barplot_donutplot_{i.lower()}.png', format = 'png', dpi = 75, bbox_inches = 'tight', transparent = True)

    plt.show()
```
