### 08.3.5. Obtendo as métricas de desempenho do modelo com os dados de teste

```python
df_comparing_metrics = pd.read_csv(consts.DATASET_COMPARING_METRICS_PANDAS)

r2_test = df_comparing_metrics.loc[0, 'Dados de Teste']

mae_test = df_comparing_metrics.loc[1, 'Dados de Teste']

rmse_test = df_comparing_metrics.loc[2, 'Dados de Teste']
```

### 08.3.6. Definindo os valores a serem simulados no modelo

```python
base_data = {
    'host_is_superhost': 't',
    'host_listings_count': 2.0,
    'property_type': 'Apartment',
    'room_type': 'Entire home/apt',
    'accommodates': 4,
    'bathrooms': 1,
    'bedrooms': 2,
    'beds': 2,
    'bed_type': 'Real Bed',
    'extra_people': 50.0,
    'minimum_nights': 2,
    'instant_bookable': 't',
    'is_business_travel_ready': 'f',
    'cancellation_policy': 'strict',
    'ano': 2018,
    'mes': 4
}

locations = {
    'Barra da Tijuca': (-23.0146345702919, -43.30877688804792),
    'Copacabana': (-22.969341672309362, -43.18236473479081),
    'Centro': (-22.913286279071503, -43.189514498918896),
    'Tijuca': (-22.930076091167322, -43.2436977670238),
    'Botafogo': (-22.950083904017422, -43.17934206190518),
    'Recreio dos Bandeirantes': (-23.025717105308974, -43.460412234411706),
}
```

### 08.3.6. Realizando a previsão com o modelo

```python
print(
    'Métricas de Desempenho:\n'
    f'- R²: {r2_test * 100:.2f}%\n'
    f'- MAE: R$ {mae_test:.2f}\n'
    f'- RMSE: R$ {rmse_test:.2f}\n'.replace('.', ',')
)

print('Previsões:')

for bairro, (lat, lon) in locations.items():

    data = pd.DataFrame({**base_data, 'latitude': [lat], 'longitude': [lon]})

    prediction = model_regression.predict(data)[0]

    value = f'{prediction:.2f}'.replace('.', ',')

    metrics = {'mae_min': prediction - mae_test, 'mae_max': prediction + mae_test, 'rmse_min': prediction - rmse_test, 'rmse_max': prediction + rmse_test}

    for i in metrics:
        
        metrics[i] = f'{metrics[i]:.2f}'.replace('.', ',')

    print(
        f'- Bairro: {bairro}\n'
        f'- Preço sugerido: R$ {value}.\n'
        f'- Intervalo esperado: R$ {metrics["mae_min"]} a R$ {metrics["mae_max"]}.\n'
        f'- Intervalo máximo: R$ {metrics["rmse_min"]} a R$ {metrics["rmse_max"]}.\n'
    )
```
