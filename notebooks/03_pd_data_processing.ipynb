## 03.3. Processamento dos dados

### 03.3.1. Removendo os registros totalmente duplicados

```python
df = df.drop_duplicates()
```

### 03.3.2. Tratando as colunas com valores nulos/vazios

```python
df = df.drop([
    'host_response_time',
    'host_response_rate',
    'security_deposit',
    'cleaning_fee',
    'review_scores_rating',
    'review_scores_accuracy',
    'review_scores_cleanliness',
    'review_scores_checkin',
    'review_scores_communication',
    'review_scores_location',
    'review_scores_value',
], axis = 1)
```

```python
df = df.dropna(subset = [
    'bathrooms',
    'bedrooms',
    'beds',
    'host_is_superhost',
    'host_listings_count',
])
```

### 03.3.3. Corrigindo os tipos de dados das colunas

```python
df['bathrooms'] = df['bathrooms'].astype('int64')

df['bedrooms'] = df['bedrooms'].astype('int64')

df['beds'] = df['beds'].astype('int64')

df['price'] = df['price'].replace({r'\$': '', ',': ''}, regex = True).astype(float)

df['extra_people'] = df['extra_people'].replace({r'\$': '', ',': ''}, regex = True).astype(float)
```

### 03.3.4. Criando colunas de tempo/idade

- **N/A**

### 03.3.5. Mesclando colunas com muitas categorias

```python
colunas_agrupar = df['property_type'].value_counts()[lambda x: x < 2000].index

df.loc[df['property_type'].isin(colunas_agrupar), 'property_type'] = 'Others'
```

```python
colunas_agrupar = df['bed_type'].value_counts()[lambda x: x < 10000].index

df.loc[df['bed_type'].isin(colunas_agrupar), 'bed_type'] = 'Others'
```

```python
colunas_agrupar = df['cancellation_policy'].value_counts()[lambda x: x < 10000].index

df.loc[df['cancellation_policy'].isin(colunas_agrupar), 'cancellation_policy'] = 'strict'
```

```python
columns = ['property_type', 'bed_type', 'cancellation_policy']

for i in columns:

    df_agg = fn_stats_pandas.groupby_count(df, i)

    print(f'- Agrupamento da coluna: {i}')

    display(HTML(df_agg.to_html(index = False)))

    fn_charts_pandas.barplot_donutplot(df_agg, i, 'Count')

    plt.savefig(f'../images/outputs/charts/pandas/nb03_barplot_donutplot_{i.lower()}.png', format = 'png', dpi = 75, bbox_inches = 'tight', transparent = True)

    plt.show()
```
