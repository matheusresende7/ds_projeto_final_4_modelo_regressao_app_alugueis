### 01.3.3. Unindo as tabelas

```python
dfs = []

meses = {'jan': 1, 'fev': 2, 'mar': 3, 'abr': 4, 'mai': 5, 'jun': 6, 'jul': 7, 'ago': 8, 'set': 9, 'out': 10, 'nov': 11, 'dez': 12}

data_folder = '../data/01_resource'

for i in os.listdir(data_folder):

    if i.endswith('.csv'):
        
        nome_mes = i[:3]
        if nome_mes not in meses: continue
        mes = meses[nome_mes]

        ano = i[-8:]
        ano = int(ano.replace('.csv', ''))
        
        arquivo_nome = os.path.join(data_folder, i)
        
        df = pd.read_csv(arquivo_nome, low_memory = False)

        df['ano'] = ano
        df['mes'] = mes

        dfs.append(df)

dataset_raw = pd.concat(dfs, ignore_index = True)
```

```python
dataset_raw.head(5)
```

```python
dataset_raw.info(verbose = True, show_counts = True)
```

### 01.3.4. Salvando a tabela unificada

```python
dataset_raw.to_csv(

    consts.DATASET_RAW_PANDAS,

    index = False

)
```
